```python
# Example: maximize f(x) = -x^2 + 4x
import random

def hill_climbing(f, x_start, iterations=100):
    current = x_start
    for i in range(iterations):
        neighbors = [current + 0.1, current - 0.1]  # small steps
        next_node = max(neighbors, key=f)
        if f(next_node) <= f(current):
            break
        current = next_node
    return current, f(current)

x_start = float(input("Enter starting value of x: "))
x_max, f_max = hill_climbing(lambda x: -x**2 + 4*x, x_start)
print(f"Maximum at x = {x_max:.2f}, f(x) = {f_max:.2f}")
```
