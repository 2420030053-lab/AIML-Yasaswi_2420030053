```python
# DFS (Depth-First Search) with user input

def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()
    visited.add(start)
    print(start, end=' ')
    for neighbor in graph.get(start, []):  # use get() to avoid key errors
        if neighbor not in visited:
            dfs(graph, neighbor, visited)

# Input graph from user
graph = {}
n = int(input("Enter number of nodes: "))
for _ in range(n):
    node = input("Enter node name: ")
    neighbors = input(f"Enter neighbors of {node} separated by space: ").split()
    graph[node] = neighbors

start = input("Enter starting node for DFS: ")

print("DFS Traversal:")
dfs(graph, start)
print()

```
