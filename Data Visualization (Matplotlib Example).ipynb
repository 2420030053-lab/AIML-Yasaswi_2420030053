```python
import matplotlib.pyplot as plt

n = int(input("Enter number of data points: "))
x = []
y = []

for i in range(n):
    xi = float(input(f"Enter x[{i}]: "))
    yi = float(input(f"Enter y[{i}]: "))
    x.append(xi)
    y.append(yi)

plt.scatter(x, y, color='blue')
plt.plot(x, y, color='red')
plt.title("Data Visualization")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
